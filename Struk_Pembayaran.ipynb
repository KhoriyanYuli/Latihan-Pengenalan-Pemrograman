```python
# Khoriyan Yuli.py
print(f"{'INPUT DATA TRANSAKSI':=^36}")

# Meminta masukan dari pengguna
nama_pelanggan = input("Nama Pelanggan: ")
total_belanja = float(input("Total Belanja Sebelum Diskon : Rp"))
persen_diskon = float(input("Persentase Diskon (%): '%'"))

# Menghitung nominal diskon dan total bayar
nominal_diskon = total_belanja * (persen_diskon / 100)
total_bayar = total_belanja - nominal_diskon

# Format mata uang rupiah dengan pemisah ribuan koma dan 2 angka desimal
str_total_awal = f"Rp {total_belanja:,.2f}"
str_diskon = f"Rp {nominal_diskon:,.2f}"
str_total_bayar = f"Rp {total_bayar:,.2f}"
str_label_diskon = f"Diskon ({persen_diskon}%):"

# Menampilkan Nota Pembayaran
print("\n" + "=" * 36)
print(f"{'STRUK PEMBAYARAN':^36}")
print(f"{nama_pelanggan.center(36)}")
print("=" * 36)
print(f"{'Total Awal':<15} : {str_total_awal:>16}")
print(f"{str_label_diskon:<15} : {str_diskon:>16}")
print("-" * 36)
print(f"{'Total Bayar':<15} : {str_total_bayar:>16}")
print("=" * 36)

```
